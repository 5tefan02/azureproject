```python
import requests

token_resp = requests.post(
    "https://login.microsoftonline.com/common/oauth2/v2.0/devicecode",
    data={
        "client_id": "2ff814a6-3304-4ab8-85cb-cd0e6f879c1d",
        "scope": "2ff814a6-3304-4ab8-85cb-cd0e6f879c1d/.default"
    }
)
print(token_resp.json()["message"])
```

```python
token_final = requests.post(
    "https://login.microsoftonline.com/common/oauth2/v2.0/token",
    data={
        "client_id": "2ff814a6-3304-4ab8-85cb-cd0e6f879c1d",
        "grant_type": "urn:ietf:params:oauth:grant-type:device_code",
        "device_code": token_resp.json()["device_code"]
    }
)
account_token = token_final.json()["access_token"]
print("Token obtinut:", account_token[:20] + "...")
```
